#### Load Pandas Library

```python
import pandas as pd
```

#### Confirming Pandas Version

```python
REQUIRED_PANDAS = "3.0.5" 
installed = pd.__version__.split("+", 1)[0]
print(f"Installed pandas version: {installed}")
if installed != REQUIRED_PANDAS:
    raise ImportError(f"Required pandas version is {REQUIRED_PANDAS}, but installed version is {installed}. Please install the correct version.")
```

#### Create Dataset URL

```python
fires_url = 'https://www.murach.com/python_analysis/RDS-2013-0009.4_SQLITE.zip'
```

#### Retrieve the Fires Dataset

```python
from urllib import request
request.urlretrieve(fires_url, filename='RDS-2013-0009.4_SQLITE.zip')
```

#### Extract the data

```python
from zipfile import ZipFile
with ZipFile('RDS-2013-0009.4_SQLITE.zip', mode='r') as zip:
    zip.extractall()
    for file in zip.infolist():
        print(file.filename, file.file_size, file.compress_size)
```

#### Running some queries

```python
import sqlite3
conn = sqlite3.connect('FPA_FOD_20170508.sqlite')
cursor = conn.cursor()
conn.cursor().execute("SELECT name FROM sqlite_master WHERE type='table'").fetchall()
```

```python
wildfires = pd.read_sql_query(
    '''SELECT fire_name, state, datetime(discovery_date) AS discovery_date, 
    datetime(cont_date) AS contained_date, fire_year, fire_size, 
    fire_size_class AS fire_class, stat_cause_descr AS fire_cause
    FROM fires''', conn)
wildfires.columns = wildfires.columns.str.lower()
wildfires.to_csv('wildfires.csv', index=False)
```
