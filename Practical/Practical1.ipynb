## Checking if first 3 columns are present

```python
# We are creating a new DataFrame called checking errors to work with finding errors
checking_errors = pd.DataFrame(raw)

first_3_column = checking_errors[["App", "Category", "Rating"]]
category_column = checking_errors["Category"]
rating_column = checking_errors["Rating"]

missing_names = first_3_column["App"].isna()
missing_category = first_3_column["Category"].isna()
missing_rating = first_3_column["Rating"].isna()
# isna() detects missing values - True mean values are absent

# True adds 1 to the sum and False adds 0
missing_names_count = missing_names.sum()
missing_category_count = missing_category.sum()
missing_rating_count = missing_rating.sum()

# Outputing all our results
display(first_3_column)                # We see that all records are false so there are no missing values
print("Missing names: ", missing_names_count)                           # sums up to 0 because there is no missing values
print("Missing categories: ", missing_category_count)
print("Missing ratings: ", missing_rating_count)

# ---------------------- Results -----------------------
# So after inspecting the names of the App columns. I have found that there are no missing values
# All are present, so we recognise that all object are there
```
